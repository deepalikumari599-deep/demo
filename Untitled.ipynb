```python
df['Category']=df['Category'].replace({'{"mode":"full"': np.nan})
```

```python
df.dropna(inplace=True)
```

```python
def wordopt(text):
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'<.*?>', '', text)
    text = re.sub(r'\[.*?\]', '', text)
    text = re.sub(f"[{re.escape(string.punctuation)}]", "", text)
    text = re.sub(r'\w*\d\w*', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
df.loc[:, 'cleaned'] = df['Message'].apply(wordopt)
```
